```python
df = df[['sender', 'subject', 'body', 'label']]
```

```python
df
```

```python
import re

def clean_text_keep_urls(text):
    if pd.isnull(text):
        return ""
    text = text.lower()
    text = re.sub(r"[^a-z0-9:/.\s@]", " ", text)  # garde urls, emails
    text = re.sub(r"\s+", " ", text).strip()
    return text

df['sender'] = df['sender'].apply(clean_text_keep_urls)
df['subject'] = df['subject'].apply(clean_text_keep_urls)
df['body'] = df['body'].apply(clean_text_keep_urls)
```

```python
df.drop_duplicates(inplace=True)
df.dropna(subset=['sender', 'subject', 'body', 'label'], inplace=True)
```

```python
spam_df = df[df.label == 1].sample(n=10000, random_state=42)
ham_df = df[df.label == 0].sample(n=10000, random_state=42)
df_balanced = pd.concat([spam_df, ham_df]).sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
df_balanced.to_csv("balanced_emails_dataset.csv", index=False)
```
